```python
from matplotlib import pyplot as pp
```

```python
import numpy as np
np.random.seed(1234)
```

```python
# Make sure that this notebook is located in the folder that contains the folders
#  - lab2_code
#  - grid_graph_search
# And the file :
#  - config.py

import path_planning
import config
```

```python
# First we generate a map

map = path_planning.generateMap2d(config.map_size)
pp.imshow(map)
```

```python
# Random Search

from grid_graph_search import RandomSearch

rs = RandomSearch( map )
path = rs.explore()
path_planning.plotMap( map, path, 'RandomSearch' )
```

```python
# BFS

from grid_graph_search import BFS

bfs = BFS( map )
path, explore_map, expand_nodes = bfs.explore()
path_planning.plotMap( map, path, 'BFS' )

print('BFS', len(path), expand_nodes )

pp.clf()
pp.matshow( explore_map )
pp.title("BFS Expansion")
```

```python
# DFS

from grid_graph_search import DFS

dfs = DFS( map )
path, explore_map, expand_nodes = dfs.explore()
path_planning.plotMap( map, path, 'DFS' )


print('DFS', len(path), expand_nodes )

pp.clf()
pp.matshow( explore_map )
pp.title("DFS Expansion")
```

```python
# Greedy Search Manhattan

from grid_graph_search import GreedySearch

gs = GreedySearch( map, distance='manhattan' )
path = gs.explore()
path_planning.plotMap( map, path, 'GS_Manhattan' ) 

print( 'GS_Manhattan', len(path) )
```

```python
# Greedy Search Euclidean

from grid_graph_search import GreedySearch

gs = GreedySearch( map, distance='euclidean' )
path = gs.explore()
path_planning.plotMap( map, path, 'GS_Euclidean' ) 

print( 'GS_Euclidean', len(path) )
```

```python
# AStar Search Manhattan

from grid_graph_search import AStar

astar = AStar( map, distance='manhattan')
path, g, h, f, expand_nodes = astar.explore()
path_planning.plotMap( map, path, 'AStar_Manhattan' ) 
```

```python
# AStar Search Euclidean

from grid_graph_search import AStar

astar = AStar( map, distance='euclidean')
path, g, h, f, expand_nodes = astar.explore()
path_planning.plotMap( map, path, 'AStar_Euclidean' ) 
```
